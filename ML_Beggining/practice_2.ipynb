```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
```

```python
data = {
    "CGPA": [7, 8, 6, 9, 8, 7, 9, 6, 8, 9,
             7.5, 8.5, 6.5, 9.2, 7.8, 8.2],
    "IQ": [110, 120, 100, 130, 125, 115, 135, 105,
           118, 140, 112, 128, 108, 138, 122, 126],
    "Communication": [6, 8, 5, 9, 8, 7, 9, 5,
                      7, 10, 7, 9, 6, 9, 8, 8],
    "Salary": [5, 8, 4, 12, 10, 7, 13, 4,
               8, 15, 6, 11, 5, 14, 9, 10]
}
df = pd.DataFrame(data)
X = df[["CGPA", "IQ", "Communication"]]
Y = df["Salary"]
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
class LinearRegressionScratch:
    def __init__(self,learning_rate=0.01,epochs=10000):
        self.lr=learning_rate
        self.epochs=epochs
        self.loss_history=[]
        self.weight=None
        self.bias=None
        self.mean=None
        self.std=None
         #Y^=(XW) + b
    def predict_scaled(self,X):
        product=np.dot(X,self.weight) + self.bias
        return product
        
        #X=(X-mean)/std
    def scaled_feature(self,X):
        self.mean=np.mean(X,axis=0)
        self.std=np.std(X,axis=0)
        self.std[self.std==0]=1
        Standardization=(X-self.mean)/self.std
        return Standardization
        
    def cost_function(self,Y,Y_pred):
        error=Y-Y_pred
        mse= np.mean(error**2)
        return mse
    
    def fit(self,X,Y):
        X=np.asarray(X,dtype=float)
        Y=np.asarray(Y,dtype=float).reshape(-1)
        X=self.scaled_feature(X)
        m,n=X.shape
        self.weight=np.zeros(n)
        self.bias= 0
        for i in range(self.epochs):
            y_pred=self.predict_scaled(X)
            error=y_pred-Y
            dw=(2/m)*np.dot(X.T,error)
            db=(2/m)*np.sum(error)
            self.weight -= self.lr* dw
            self.bias -= self.lr* db
            loss=self.cost_function(Y,self.predict_scaled(X))
            self.loss_history.append(loss)
        
        print("Maximum loss = ",loss)
        return self
    def predict(self,X):
        X=np.asarray(X, dtype=float)
        X_scaled=(X-self.mean)/self.std
        return self.predict_scaled(X_scaled)
            
```

```python
model=LinearRegressionScratch(epochs=10000)
y_pred=model.fit(X_train,Y_train).predict_scaled(X_test)
#y_pred=model.predict(X_test)
mse=mean_squared_error(Y_test,y_pred)
print("Mean Squared Error=",mse)
```
