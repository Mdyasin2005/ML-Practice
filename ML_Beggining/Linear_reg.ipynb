```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report,mean_squared_error
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv('placement.csv')
X=df[['cgpa']]
y=df['package']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
class LinearRegressionScratch:
    def __init__(self,epochs=10000,learning_rate=0.01):
        self.epochs= epochs
        self.learning_rate=learning_rate
        self.loss_history=[]
        self.m=0
        self.b=0
    def lrloss(self,y,y_pred):
        return np.mean((y-y_pred)**2)  
    
    def fit(self,x,y):
        n_samples,n_features = np.shape(x)
        self.m=np.zeros(n_features)
        self.b=0
        for epoch in range(self.epochs):
            y_pred=np.dot(x,self.m)+self.b
            loss=self.lrloss(y,y_pred)
            self.loss_history.append(loss)
            mean_x=np.mean(x)
            mean_y=np.mean(y)
            
            dm=(-2/n_samples)*np.dot(x.T,(y-y_pred))
            db=(-2/n_samples)*np.sum(y-y_pred)
            
            self.m= self.m - self.learning_rate*dm
            self.b=self.b - self.learning_rate*db
            
            if epoch % 1000==0:
                print(f"Epoch{epoch:5d}: loss={loss:.6f}")
        return self
    
    def predict(self,x):
        x = np.array(x)
       
        return np.dot(x,self.m)+self.b
    
```

```python
lr=LinearRegressionScratch(epochs=10000)
y_pred=lr.fit(X_train,y_train).predict(X_test)
#print(y_pred)
mse=mean_squared_error(y_test,y_pred)
```
